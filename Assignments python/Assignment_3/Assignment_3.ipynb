```python
'''"Application Data"1. Create a class BankAccount with the following:
a) Attributes: account_holder, account_number, and balance.
b) Methods: deposit(amount), withdraw(amount), and get_balance().
c) withdraw() should print a message and do nothing if the amount exceeds the current balance.
d) A __str__ method that returns a readable summary of the account.
Create two BankAccount objects and perform a few deposits and withdrawals on each.
''';
```

```python
class BankAccount:
    def __init__(self,account_holder,account_number, balance=0):
        self.name = account_holder
        self.num = account_number
        self.bal = balance

    def deposit(self, amt):
        if amt > 0:
            self.bal += amt
            print("Deposit:", amt)
        else:
            print(f"amount is less than 0 ! can't deposit")

    def withdraw(self, amt):
        if amt > self.bal:
            print("No money")
        else:
            self.bal -= amt
            print("Withdraw:", amt)

    def get_balance(self):
        return self.bal

    def __str__(self):
        return f"account holder->{self.name} account number->{self.num} Bal:{self.bal}"

# test
```

```python
'''
2. Create a class Student with attributes name, roll_no, and marks (a list of 5 subject marks). Add
methods:
a) calculate_percentage() — returns the average of all marks.
b) get_grade() — returns 'A' for >= 80, 'B' for >= 60, 'C' for >= 40, and 'Fail' otherwise.
c) display() — prints all details including percentage and grade.
Create a list of 3 Student objects and call display() on each.
''';
```

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def calculate_percentage(self):
        return sum(self.marks) / 5

    def get_grade(self):
        p = self.calculate_percentage()
        if p >= 80:
            return 'A'
        if p >= 60:
            return 'B'
        if p >= 40:
            return 'C'
        return 'Fail'

    def display(self):
        print(f"name->{self.name},roll_no->{self.roll_no}, marks=>{self.marks}")
        print("Percent:", self.calculate_percentage())
        print("Grade:", self.get_grade())

```

```python
'''
3. Create a class Employee with private attributes __name, __emp_id, and __salary. Add:
a) Getter and setter methods for all three attributes.
b) The salary setter should raise a ValueError if the value is below 10000.
c) A method give_raise(percent) that increases salary by the given percentage.
Demonstrate that the private attributes cannot be accessed directly, and that all access goes
through the methods.
''';
```

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.__name = name
        self.__emp_id = emp_id
        if salary < 10000:
            raise ValueError("Salary must be >= 10000")
        self.__salary = salary

    def get_name(self):
        return self.__name
    def set_name(self, name):
        self.__name = name

    def get_emp_id(self):
        return self.__emp_id
    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id

    def get_salary(self):
        return self.__salary
    def set_salary(self, salary):
        if salary < 10000:
            raise ValueError("Salary must be >= 10000")
        self.__salary = salary

    def give_raise(self, percent):
        self.__salary += self.__salary * percent / 100

    def show(self):
        print(self.__name, self.__emp_id, self.__salary)

```

```python
'''
4. Build a class SecureVault that stores a private __pin and a private __data dictionary. It should
have:
a) A method add_entry(pin, key, value) that adds to __data only if the pin matches.
b) A method get_entry(pin, key) that returns the value only if the pin matches.
c) Any wrong pin attempt should print 'Access Denied' and return nothing
''';
```

```python
class SecureVault:
    def __init__(self, pin):
        self.__pin = pin
        self.__data = {}

    def add_entry(self, pin, key, value):
        if pin == self.__pin:
            self.__data[key] = value
            print("Added")
        else:
            print("Access Denied")

    def get_entry(self, pin, key):
        if pin == self.__pin:
            return self.__data.get(key)
        else:
            print("Access Denied")
```

Section C - Inheritance and method overriding

```python
'''
5. Create a base class Shape with a method area() that returns 0 and a method describe() that
prints 'I am a shape'. Create three subclasses:
a) Circle(radius) — overrides area() using pi * r^2.
b) Rectangle(length, width) — overrides area().
c) Triangle(base, height) — overrides area().
Each subclass should also override describe() to print something specific about itself. Create one
object of each and call both methods.
''';
```

```python
class Shape:
    def area(self):
        return 0
    
    def describe(self):
        print("I am a shape")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius * self.radius
    
    def describe(self):
        print(f"I am a circle with radius {self.radius}")

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.length * self.width
    
    def describe(self):
        print(f"I am a rectangle {self.length}x{self.width}")

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def area(self):
        return 0.5 * self.base * self.height
    
    def describe(self):
        print(f"I am a triangle with base {self.base} and height {self.height}")

```

```python
'''
6. Create a class Vehicle with attributes make, model, and year and a method start_engine() that
prints 'Engine started'. Create two subclasses:
a) Car(make, model, year, num_doors) — overrides start_engine() to include the car's name in
the message.
b) ElectricCar(make, model, year, battery_capacity) — overrides start_engine() to print 'Silent
start — battery at X kWh'.
Use super() in both subclasses to call the parent __init__. Create objects and test all methods.
''';
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def start_engine(self):
        print("Engine started")

class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year) 
        self.num_doors = num_doors
    
    def start_engine(self):
        print(f"{self.make} {self.model} engine started")

class ElectricCar(Vehicle):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year) 
        self.battery_capacity = battery_capacity
    
    def start_engine(self):
        print(f"Silent start — battery at {self.battery_capacity} kWh")

```

```python
v = Vehicle("Generic", "V1", 2020)
c = Car("Honda", "City", 2023, 4)
e = ElectricCar("Tesla", "Model 3", 2024, 75)

```

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

    def multiply(self, a, b, c=1):
        return a * b * c

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Vector(new_x, new_y)

    def display(self):
        print("Vector(", self.x, ",", self.y, ")")


```

```python
'''
9. Perform the following file operations. Use plain text files only.
a) Write a list of 5 student names to a file students.txt, one name per line.
b) Read the file back and print each name with its line number.
c) Append two more names to the same file without overwriting existing content.
d) Read the final file and count the total number of students.
e) Copy the contents of students.txt into a new file students_backup.txt.
''';
```

```python
# a) Write 5 student names to students.txt
students = ["Amit", "Priya", "Rohit", "Sneha", "Vikram"]

f= open("students.txt", "w")
for name in students:
        f.write(name + "\n")

print("Part a: File created with 5 names")

# b) Read file and print each name with line number - Simple logic
print("\nPart b: Reading file")
f= open("students.txt", "r")
lines = f.readlines()

for i in range(len(lines)):
    print(f"{i+1}: {lines[i].strip()}")

# c) Append two more names without overwriting
f= open("students.txt", "a")
f.write("Arjun\n")
f.write("Kavya\n")

print("\nPart c: Added 2 more names")

# d) Read final file and count total students
f= open("students.txt", "r")
total = len(f.readlines())
    
print(f"\nPart d: Total students = {total}")

# e) Copy contents to students_backup.txt
src=open("students.txt", "r")
dest=open("students_backup.txt", "w")
dest.write(src.read())

print("Part e: Backup file created")
```

Section F- Menu Driven With OOP

```python
'''
10. Build a menu-driven Student Record Manager using OOP and file handling. The app should
have a class StudentManager with the following menu options:
1. Add Student — takes name, roll number, and marks for 3 subjects.
2. View All — displays all stored students with their average marks.
3. Search Student — finds a student by roll number.
4. Delete Student — removes a student by roll number.
5. Save to File — writes all records to students_data.txt.
6. Load from File — reads records back from students_data.txt on startup.
7. Exit
Requirements: use a class with a list as internal storage, handle invalid menu choices gracefully,
and use a while loop to keep the menu running until the user chooses to exit.
''';
```

```python
class StudentManager:
    def __init__(self):
        self.students = []
        self.load_from_file()

    def add_student(self):
        name = input("Enter Name: ")
        roll = input("Enter Roll Number: ")

        m1 = float(input("Enter Marks of Subject 1: "))
        m2 = float(input("Enter Marks of Subject 2: "))
        m3 = float(input("Enter Marks of Subject 3: "))

        student = [roll, name, m1, m2, m3]
        self.students.append(student)

        print("Student Added Successfully")

    def view_all(self):
        if len(self.students) == 0:
            print("No Records Found")
            return

        for student in self.students:
            avg = (student[2] + student[3] + student[4]) / 3

            print("\nRoll No:", student[0])
            print("Name:", student[1])
            print("Marks:", student[2], student[3], student[4])
            print("Average:", avg)

    def search_student(self):
        roll = input("Enter Roll Number to Search: ")

        found = False

        for student in self.students:
            if student[0] == roll:
                avg = (student[2] + student[3] + student[4]) / 3

                print("\nStudent Found")
                print("Roll No:", student[0])
                print("Name:", student[1])
                print("Average:", avg)

                found = True
                break

        if found == False:
            print("Student Not Found")

    def delete_student(self):
        roll = input("Enter Roll Number to Delete: ")

        found = False

        for student in self.students:
            if student[0] == roll:
                self.students.remove(student)
                print("Student Deleted Successfully")
                found = True
                break

        if found == False:
            print("Student Not Found")

    def save_to_file(self):
        file = open("students_data1.txt", "w")

        for student in self.students:
            data = str(student[0]) + "," + str(student[1]) + "," + \
                   str(student[2]) + "," + str(student[3]) + "," + \
                   str(student[4]) + "\n"

            file.write(data)

        file.close()
        print("Data Saved Successfully")

    def load_from_file(self):
        file = open("students_data1.txt", "r")

        for line in file:
            data = line.strip().split(",")

            roll = data[0]
            name = data[1]
            m1 = float(data[2])
            m2 = float(data[3])
            m3 = float(data[4])

            self.students.append([roll, name, m1, m2, m3])

        file.close()


# Create file if it doesn't exist
file = open("students_data1.txt", "a")
file.close()

# Create object
manager = StudentManager()

# Menu
while True:
    print("\n===== Student Record Manager =====")
    print("1. Add Student")
    print("2. View All")
    print("3. Search Student")
    print("4. Delete Student")
    print("5. Save To File")
    print("6. Load From File")
    print("7. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        manager.add_student()

    elif choice == "2":
        manager.view_all()

    elif choice == "3":
        manager.search_student()

    elif choice == "4":
        manager.delete_student()

    elif choice == "5":
        manager.save_to_file()

    elif choice == "6":
        manager.students.clear()   # Prevent duplicate loading
        manager.load_from_file()
        print("Data Loaded Successfully")

    elif choice == "7":
        manager.save_to_file()
        print("Program Ended")
        break

    else:
        print("Invalid Choice")
```

Section A — Exception Handling

```python
'''
1. Write a function safe_divide(a, b) that divides two numbers and handles ZeroDivisionError and
TypeError separately, printing an appropriate message for each. Use a finally block to print
'Operation attempted.' regardless of the outcome. Test it with valid numbers, division by zero, and a
string input.
''';
```

```python
def safe_divide(a, b):
    try:
        result = a / b
        print("Result:", result)

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")

    except TypeError:
        print("Error: Invalid data type. Please enter numbers only.")

    finally:
        print("Operation attempted.\n")

```

```python
safe_divide(12,2)
```

```python
safe_divide(10,0)
```

```python
safe_divide('10','1')
```

```python
'''
2. Write a function validate_age(age) that raises a custom exception InvalidAgeError (a class you
define, inheriting from Exception) if age is negative or greater than 120. Catch this exception in a
calling block and print a user-friendly message. Test it with at least 3 different values.
''';
```

```python

class InvalidAgeError(Exception):
    pass


def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError("Invalid age entered.")
    print("Valid Age:", age)

```

```python
try:
    validate_age(25)
except InvalidAgeError as a:
    print('error',a)
```

```python
try:
    validate_age(-10)
except InvalidAgeError as a:
    print('error',a)
```

```python
try:
    validate_age(125)
except InvalidAgeError as a:
    print('error',a)
```

```python
'''
4. Create a 1D NumPy array of integers from 1 to 20. Perform the following:
a) Reshape it into a 4x5 2D array.
b) Print the sum of each row and each column.
c) Extract all elements greater than 10 using boolean indexing.
d) Replace all even numbers in the array with -1

''';
```

```python
import numpy as np

# Create 1D array from 1 to 20
arr = np.arange(1, 21)
print("Original 1D array:\n", arr)

# a) Reshape into 4x5 array
arr_2d = arr.reshape(4, 5)
print("\nReshaped 4x5 array:\n", arr_2d)

# b) Sum of each row and each column
row_sum = arr_2d.sum(axis=1)
col_sum = arr_2d.sum(axis=0)

print("\nSum of each row:", row_sum)
print("Sum of each column:", col_sum)

# c) Elements greater than 10
greater_than_10 = arr_2d[arr_2d > 10]
print("\nElements greater than 10:", greater_than_10)

# d) Replace even numbers with -1
modified_arr = arr_2d.copy()
modified_arr[modified_arr % 2 == 0] = -1

print("\nArray after replacing even numbers with -1:\n", modified_arr)
```

```python
'''
6. Generate a NumPy array of 15 random integers between 1 and 100 using np.random.randint()
(set a seed of 42 for reproducibility). Then:
a) Find the mean, median, standard deviation, min, and max.
b) Sort the array in descending order.
c) Find the indices of the 3 largest values using np.argsort().
d) Normalize the array so all values lie between 0 and 1.
''';
```

```python
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Generate array
arr = np.random.randint(1, 101, 15)
print("Original Array:\n", arr)

# a) Statistics
print("\nMean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))

# b) Sort in descending order
sorted_desc = np.sort(arr)[::-1]
print("\nSorted (Descending):\n", sorted_desc)

# c) SIMPLE argsort logic (TOP 3 LARGEST VALUES)
sorted_indices = np.argsort(arr)     # ascending order indices
top3_indices = sorted_indices[-3:]   # last 3 = largest values

print("\nIndices of 3 largest values:", top3_indices)
print("Top 3 values:", arr[top3_indices])

```

```python
'''
7. Create a Pandas DataFrame from the dictionary below and perform the following:
data = { "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"], "Age": [23, 21,
25, 22, 24], "Department": ["AI", "Data Science", "AI", "Web Dev", "Data
Science"], "Score": [88, 92, 79, 95, 67], }
a) Display the first 3 rows and the data types of each column.
b) Add a new column Grade based on Score (>=90 'A', >=75 'B', else 'C').
c) Sort the DataFrame by Score in descending order.
d) Filter rows where Department is 'AI'.
''';
```

```python
import pandas as pd

#
data = {
    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
    "Age": [23, 21, 25, 22, 24],
    "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
    "Score": [88, 92, 79, 95, 67]
}


df = pd.DataFrame(data)

print("Original DataFrame:\n", df)


print("\nFirst 3 Rows:\n", df.head(3))
print("\nData Types:\n", df.dtypes)

df["Grade"] = ""

df.loc[df["Score"] >= 90, "Grade"] = "A"
df.loc[(df["Score"] >= 75) & (df["Score"] < 90), "Grade"] = "B"
df.loc[df["Score"] < 75, "Grade"] = "C"

print("\nDataFrame with Grades:\n", df)


sorted_df = df.sort_values(by="Score", ascending=False)
print("\nSorted by Score (Descending):\n", sorted_df)

# d) Filter Department = AI
ai_df = df[df["Department"] == "AI"]
print("\nAI Department Students:\n", ai_df)
```

```python
'''
8. Using the same DataFrame from Q7:
a) Group by Department and find the average Score for each.
b) Find the row with the highest Score using idxmax().
c) Rename the column Score to Marks.
d) Drop the Age column and display the resulting DataFrame
''';
```

```python
import pandas as pd

data = {
    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
    "Age": [23, 21, 25, 22, 24],
    "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
    "Score": [88, 92, 79, 95, 67]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df)


dept_avg = df.groupby("Department")["Score"].mean()
print("\nAverage Score by Department:\n", dept_avg)


max_index = df["Score"].idxmax()
highest_score_row = df.loc[max_index]
print("\nRow with Highest Score:\n", highest_score_row)

df = df.rename(columns={"Score": "Marks"})
print("\nAfter Renaming Score to Marks:\n", df)


df = df.drop(columns=["Age"])
print("\nAfter Dropping Age Column:\n", df)
```

```python
'''9. Create a DataFrame with some missing values as shown below and handle them:
import numpy as np data = { "Product": ["Pen", "Notebook", "Eraser",
"Sharpener", "Ruler"], "Price": [10, np.nan, 5, 8, np.nan], "Stock": [100, 50,
np.nan, 30, 20], }
a) Check the number of missing values in each column using isnull().sum().
b) Fill missing Price values with the column mean.
c) Drop rows where Stock is still missing.
d) Save the cleaned DataFrame to a CSV file named cleaned_products.csv without the index.
''';
```

```python
import pandas as pd
import numpy as np


data = {
    "Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
    "Price": [10, np.nan, 5, 8, np.nan],
    "Stock": [100, 50, np.nan, 30, 20]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df)


print("\nMissing values in each column:\n", df.isnull().sum())


price_mean = df["Price"].mean()
df["Price"] = df["Price"].fillna(price_mean)

print("\nAfter filling missing Price:\n", df)


df = df.dropna(subset=["Stock"])

print("\nAfter dropping rows with missing Stock:\n", df)

df.to_csv("cleaned_products.csv", index=False)

print("\nCleaned data saved to 'cleaned_products.csv'")
```

```python
#read purchase cosrt  and selling cost from the user
# state profit or loss and how much
```

```python
sellingprice = float(input('enter selling price'))
purchasingprice =float(input('enter purchasing price'))

if sellingprice > purchasingprice:
    profit=sellingprice-purchasingprice
    print('profit',profit)

elif purchasingprice > sellingprice:
    loss=purchasingprice-sellingprice
    print('loss',loss)

else:
    print('no loss no profit')
```

```python
"""
menu driven ->
take a string from the user
menu-> 1 upper,2 lowercase,3 replace a particlar part of the string
for third option it should which part to replace with what

"""
```

```python
text=input('enter the string')

print('menu\n1.upper\n2.lowercse\n3.replace part')

ch=int(input('enter choice'))
if ch==1:
    result=text.upper()
    print('uppercase',result)

elif ch==2:
    result=text.lower()
    print("loweecse",result)

elif ch==3:
    old=input('enter the replace')
    new=input('enter what replace')
    result=text.replace(old,new)
    print(f"after replaced {result}")

else:
    print('invalid choice')
```

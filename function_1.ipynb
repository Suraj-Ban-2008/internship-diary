```python
#create the function to find factorial of number
```

```python
def fact(n):
    f=1
    for i in range(1,n+1):
        f=f*i    
    print(f) 
    
    
```

```python
fact(5)
```

```python
fact(10)
```

```python
def addall(*args):
    add=0
    for i in (args):
        add+=i
    return add
    
```

```python
addall(1,2,3,4,5)
```

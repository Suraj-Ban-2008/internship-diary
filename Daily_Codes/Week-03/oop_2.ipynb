```python
'''
create a class called student 
create 2-3 objects havving attribute name,age

create a method which will add askill (eg.python,ml,java) to the student 
where you can view the info of it later (i.e.skills learned by a student)
''';
```

```python
class student:
    def __init__(self, name, age):
        self.name=name
        self.age=age
        self.skills=[]

    def add(self,skill):
        self.skills.append(skill)
        print(skill,'added')
        
    def info(self):
        print(f"name=>{self.name},age=>{self.age}, skill=>{self.skills}")
```

```python
s1=student("ali",20)
s1.add('python')
s1.add('ML')
s1.info()
```

```python
s2=student('jay',23)
s2.info()
```

```python
s3=student('ram',19)
s3.add('java')
s3.add('ML')
s3.info()
```

## inheritence

```python
class Person:
    insti='mitu'

    def __init__(self, age, ht):
        self.age=age
        self.ht=ht

    def show(self):
        print(f"age->{self.age}, height->{self.ht},institue->{self.insti}")
```

```python
class Ravi(Person):
    def __init__(self, age, ht, sal):
        super().__init__(age,ht)
        self.sal=sal

    def show_sal(self):
        print(f"salary is{self.sal}")
        
```

```python
a=Ravi(20,6.1,35000)
```

```python
a.show_sal()
```

```python
a.show()
```

```python
class menuapp:

    def __init__(self):
        self.itemlist=[]

    def options(self):
        print('menu\n 1.add items\n2.view items\n3.exit')

    def add_items(self):
        item=input('enter item to add in list')
        self.itemlist.append(item)

    def view_item(self):
        return self.itemlist

    def run(self):
        while True:
            self.options()
            ch=int(input('enter your choice'))


            if ch==1:
                self.add_items()
            elif ch==2:
                print(self.view_item())
            elif ch==3:
                break
            else:
                print('invalid choice')
```

```python
m=menuapp()
```

```python
m.run()
```

```python
class bank:
    
    def __init__(self,balance):
        self.balance=balance
    def options(self):
        
        print("menu\n1.check \n2.add amount\n3.withdraw\n4.exit")
    def check(self):
        print('current balance is',self.balance)
        
    def add(self):
        rupees=float(input('enter to add rupees'))
        self.balance+=rupees
        print(self.balance)

    def withdraw(self):
        amount=float(input('enter amount to withdraw'))
        if amount>self.balance:
            print('insuffiecient balance')
        else:
            self.balance-=amount
            print(self.balance)

    def run(self):
        while True:
            self.options()
            ch=int(input('enter your choice'))


            if ch==1:
                self.check()

            elif ch==2:
                self.add()
            elif ch==3:
                self.withdraw
            elif ch==4:
                break
            else:
                print('invlid choice')

            
            
```

```python
b=bank(20000)
```

```python
b.run()
```

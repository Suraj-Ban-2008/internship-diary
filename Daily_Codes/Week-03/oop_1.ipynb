```python
class Person:
    def __init__(suraj, name,age,height):
        suraj.name=name
        suraj.__age=age     #private variable
        suraj.height=height
    def show_info(self):
        return f"name->{self.name}, age->{self.__age}, ht->{self.height}"
```

```python
p1=Person('jatin',20,5.8)
p2=Person('sachin',22,5.9)
```

```python
p2.show_info()
```

```python
'''
create  a class called Employee=>
attiribute->company,name,salary,job_designation 
here all the objects of this company will have same common company
make salary as private variable 

create 3 object->e1,e2,e3 with given attribute

method->1.method to calculate the anual salary
2.method to show entire data
''';
```

```python
class Employee:
    company='Tata'
    def __init__(self,name,monthly_salary,job_designation):
        self.name=name
        self.__monthly_salary=monthly_salary
        self.job_designation=job_designation

    def anual_sal(self):
        print(self)
        print(f"anual salary=>{self.__monthly_salary*12}")
    def show_all(self):
        return f"company=>{self.company},name=>{self.name},monthly_salary=>{self.__monthly_salary},job destination=>{self.job_designation}"
    
```

```python
e1=Employee('ram',10000,'fresher')
e2=Employee('sham',20000,'HR')
e3=Employee('vikas',12000,'manager')
```

```python
e1.anual_sal()
```

```python
e2.anual_sal()
```

```python
e3.anual_sal()
```

```python
e1.show_all()
```

```python
e2.show_all()
```

```python
e3.show_all()
```

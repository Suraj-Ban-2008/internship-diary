```python
'''
take a string as the user input
use loop and if else to check if a specific character is present in string or not.
get output for both if the charcter is present and if is not present.
take 2 chacters ot cheeck in the string

user input ->string ,character 1,character2


'''
```

```python
s3=input('entrthe string')

ch1=input('enter the first character ')
ch2=input('enter the ssecond character ')

found_ch1=False
found_ch2=False

for ch in s3:
    if ch==ch1:
        found_ch1=True
    if ch==ch2:
        found_ch2=True

if found_ch1:
    print(f"the character {ch1} is present in string")
else:
    print(f"the character {ch1} is not present present in string")


if found_ch2:
    print(f"the character {ch2} is present in string")
else:
    print(f"the character {ch2} is not present in string")
```

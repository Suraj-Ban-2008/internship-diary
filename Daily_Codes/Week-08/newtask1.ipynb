```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
x=df.drop('price_range',axis=1)
y=df['price_range']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=0,test_size=0.2)
```

```python
scaler=MinMaxScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
print(x_train_scaled.shape)
print(x_test_scaled.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. RANDOM FOREST
rf = RandomForestClassifier(random_state=0)
rf.fit(x_train_scaled, y_train)
y_pred_rf = rf.predict(x_test_scaled)
acc_rf = accuracy_score(y_test, y_pred_rf)
print("1. Random Forest Accuracy:",acc_rf)

# 2. INDIVIDUAL KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_scaled, y_train)
y_pred_knn = knn.predict(x_test_scaled)
acc_knn = accuracy_score(y_test, y_pred_knn)
print("2. Individual KNN Accuracy:",acc_knn)

# 3. BAGGING OF KNN with 50 estimators
bag_knn = BaggingClassifier(
    estimator=KNeighborsClassifier(n_neighbors=5),
    n_estimators=50, 
    random_state=0
)
bag_knn.fit(x_train_scaled, y_train)
y_pred_bag = bag_knn.predict(x_test_scaled)
acc_bag = accuracy_score(y_test, y_pred_bag)
print("3. Bagging KNN Accuracy:",acc_bag)
```

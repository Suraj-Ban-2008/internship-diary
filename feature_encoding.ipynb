```python
""""

load the taxis dataset from seaborn 
remove pickup and dropoff columns from the dataset
encode all the catrgorical column using lable encoding
ideally create a function which encode when u pass the dataframe 
""";
```

```python
df=df.drop(['pickup','dropoff'],axis=1)
```

```python
def encode_column(dataframe):
    df_encoded=dataframe.copy()
    le=LabelEncoder()



    for col in df_encoded.columns:
        if df_encoded[col].dtype=='object':
            df_encoded[col]=le.fit_transform(df_encoded[col])
    return df_encoded
```

```python
df_encoded=encode_column(df)
```

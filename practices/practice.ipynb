```python
# Q.3 manual list reversal
l2=['apple','banana','cherry','date','elderberry']
```

```python
reverse=[]
for i in range(len(l2) -1,-1,-1):
    reverse.append(l2[i])
    

print(reverse)
```

```python
# Q.7 list palindrome check
li=[1,2,3,2,1]
li1=['a','b','c','b','d']
```

```python
li_rev=li.copy()
li1_rev=li1.copy()
li.reverse()
li1.reverse()
if li==li_rev:
    print('first palindrome')
else:
    print('not palindrome first')
if li1==li1_rev:
    print('second palindrome')
else:
    print('not palindrome second')
```

```python
# Q.9 flatten a nested list
nested_list=[1,[2,3],4,[5,6,7],8]
```

```python
flatten=[]
for i in nested_list:
    if type(i)==list:
        for x in i:
            flatten.append(x)
    else:
        flatten.append(i)
print(flatten)

```

```python
#Q.10 find common elements

p=[1,2,3,4,5]
q=[4,5,6,7,8]
```

```python
new_list=[]
for i in p:
    for j in q:
        if i==j:
            new_list.append(i)
```

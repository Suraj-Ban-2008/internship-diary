```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([10,20,30,40,50,60,70])
y=np.array([11,22,33,44,55,66,77])
z=np.array([45,32,45,67,34,23,89])
```

Line Chart

```python
plt.plot(x,y);
```

```python
plt.xlabel('x values')
plt.ylabel('y values')
plt.title('line chart: x vs y')
plt.plot(x,y,marker='o',mfc='red',ls='--');
plt.grid();

# 0,*,H,<,>,^
```

```python
plt.xlabel('x values')
plt.ylabel('z values')
plt.title('line chart: x vs z')
plt.plot(x,z,marker='o',mfc='blue',ls='--');
plt.grid();
```

```python
plt.xlabel('x values')
plt.ylabel('z values')
plt.title('line chart: x vs z')
plt.plot(x,z,marker='o',mfc='blue',ls='--',label='2024');
plt.plot(x,y,marker='o',mfc='red',ls='--',label='2026');
plt.grid();
```

```python
plt.figure(figsize=(16,9))
plt.subplot(2,2,1)
plt.xlabel('x values')
plt.ylabel('y values')
plt.title('line chart: x vs y')
plt.plot(x,y,marker='o',mfc='red',ls='--');
plt.grid();

plt.subplot(2,2,2)
plt.xlabel('x values')
plt.ylabel('z values')
plt.title('line chart: x vs z')
plt.plot(x,z,marker='o',mfc='blue',ls='--');
plt.grid();
```

Bar Chart

```python
courses=['python','lisp','c','r','php']
students=[68,30,47,39,60]
```

```python
plt.bar(courses,students,color='cyan',edgecolor='red');
plt.xlabel('programming courses')
plt.ylabel('number of students')
plt.title('students enrolled per course')
plt.grid()
```

```python
#load-->social_network_ads.csv
#plot bar graph showing the count of how many people purchased and how many did not
# purchased column ==>0 purchased, 1-->not purchased
```

```python
ads=pd.read_csv('Social_Network_Ads.csv')
```

```python
ads
```

```python
cnt=ads['Purchased'].value_counts()
```

```python
plt.bar(['purchased','not purchased'],cnt.values,color=['green','red'])
plt.xlabel('status')
plt.ylabel('number of peple')
plt.title('purchased count from Social network ads')
plt.grid()
```

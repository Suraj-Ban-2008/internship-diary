```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([10,20,30,40,50,60,70])
y=np.array([11,22,33,44,55,66,77])
z=np.array([45,32,45,67,34,23,89])
```

Line Chart

```python
plt.plot(x,y);
```

```python
plt.xlabel('x values')
plt.ylabel('y values')
plt.title('line chart: x vs y')
plt.plot(x,y,marker='o',mfc='red',ls='--');
plt.grid();

# 0,*,H,<,>,^
```

```python
plt.xlabel('x values')
plt.ylabel('z values')
plt.title('line chart: x vs z')
plt.plot(x,z,marker='o',mfc='blue',ls='--');
plt.grid();
```
